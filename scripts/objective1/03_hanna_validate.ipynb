```python
### imports
import numpy as np
import pandas as pd
from pathlib import Path
from scipy.stats import spearmanr
```

```python
review_scores=pd.read_csv("review_scores.csv")
hanna_scores=pd.read_csv("hanna_scores.csv")
```

```python
mapping={
'Narrative Structure & Quality':'Coherence',
'Character & Emotion':'Empathy',
'Originality':'Surprise',
'Immersion':'Engagement',
'Thematic Depth':'Complexity',
'Writing Style':'Relevance'
}
```

```python
results=[]

for c,h in mapping.items():
    rho,_=spearmanr(review_scores[c], hanna_scores[h])
    results.append([c,h,rho])
corr_df=pd.DataFrame(results, columns=['custom','hanna','rho'])
```
